# Extração dos mapas das facções

Este notebook existe para executar o script `05_extrair_mapas_faccoes.py` em
ambientes como o Google Colab, sem duplicar a lógica principal do projeto.

Antes de rodar no Colab, o diretório `TCC` precisa estar
disponível no ambiente, por exemplo em `/content/TCC`,
`/content/TCC` ou em uma pasta do Google Drive montada
no Colab.

Se a detecção automática falhar, preencha a variável `CAMINHO_PROJETO` na
célula de localização do projeto.


```python
from pathlib import Path
import subprocess
import sys

EM_COLAB = "google.colab" in sys.modules

if EM_COLAB:
    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "playwright",
            "geopandas",
            "pyproj",
            "shapely"
        ],
        check=True,
    )
    subprocess.run(
        [sys.executable, "-m", "playwright", "install", "chromium"],
        check=True,
    )

print("Dependencias prontas.")
```

```python
from pathlib import Path

# Se necessario, preencha manualmente com o caminho da pasta
# TCC no Colab ou no Drive montado.
CAMINHO_PROJETO = ""
# Exemplo:
# CAMINHO_PROJETO = "/content/drive/MyDrive/TCC"

candidatos = []
if CAMINHO_PROJETO:
    candidatos.append(Path(CAMINHO_PROJETO))

candidatos.extend([
    Path.cwd(),
    Path.cwd().parent,
    Path.cwd() / "TCC",
    Path.cwd() / "TCC" / "TCC",
    Path("/content/TCC"),
    Path("/content/TCC"),
    Path("/content/drive/MyDrive/TCC"),
    Path("/content/drive/MyDrive/TCC"),
])

projeto = None
testados = []

for candidato in candidatos:
    candidato = candidato.expanduser()
    script = candidato / "01_preparacao_dados" / "05_extrair_mapas_faccoes.py"
    testados.append(str(candidato))
    if script.exists():
        projeto = candidato.resolve()
        break

if projeto is None:
    raise FileNotFoundError(
        "Nao foi possivel localizar a pasta TCC. "
        "Defina CAMINHO_PROJETO com o caminho correto. "
        "Caminhos testados: " + " | ".join(testados)
    )

print("Projeto localizado em:", projeto)
```
